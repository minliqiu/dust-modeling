```python
def get_unique_elements(df):
    m_Planet_eff_unique = df['m_Planet_eff'].unique()
    # m_Star/m_Sun_unique = df['m_Star/m_Sun'].unique()
    m_Planet_unique = df['m_Planet/m_J'].unique() # in m_J
    a_p_unique = df['a_p/R_Sun'].unique() # in R_Sun
    beta_unique = df['beta'].unique()
    inc_unique = df['inc_initial'].unique()
    
    print ('m_Planet_eff\n', m_Planet_eff_unique)
    print ('m_Planet/m_J\n', m_Planet_unique)
    print ('a_p/R_Sun\n', a_p_unique)
    print ('beta\n', beta_unique)
    print ('inc_initial\n', inc_unique)
    
    return m_Planet_eff_unique, m_Planet_unique, a_p_unique, beta_unique, inc_unique
```

```python
fate_array = ['sublimation', 'collision', 'ejection', 'incomplete']
fate_colors = {
    'sublimation': 'r',
    'collision': 'b',
    'ejection': 'c',
    'incomplete': 'y'}
    
```

```python
def plot_group_by_one(df, group_param='beta'):
    # Get all unique parameter values
    m_Planet_eff_unique, m_Planet_unique, a_p_unique, beta_unique, inc_unique = get_unique_elements(df)

    # Loop over all other parameters except the one we're grouping by
    for combo in product(m_Planet_eff_unique, m_Planet_unique, a_p_unique, inc_unique):
        eff, mp, ap, inc = combo
        
        # Filter DataFrame for this unique combination
        subdf = df[
            (df['m_Planet_eff'] == eff) &
            (df['m_Planet/m_J'] == mp) &
            (df['a_p/R_Sun'] == ap) &
            (df['inc_initial'] == inc)
        ]

        if len(subdf) == 0:
            continue  # skip empty combinations

        # Group by the selected parameter (e.g. beta)
        grouped = subdf.groupby(group_param)
        x_vals = []
        y_vals = {fate: [] for fate in fate_array}

        for name, group in grouped:
            total = len(group)
            if total == 0:
                continue
            x_vals.append(name)
            for fate in fate_array:
                count = len(group[group['final_fate'] == fate])
                y_vals[fate].append(count / total)

        # Plot
        fig = plt.figure(figsize=(8, 5))
        ax = plt.subplot(111)
        ax.set_xscale('log')
        for fate in fate_array:
            if y_vals[fate]:  # only plot if there is data
                plt.plot(x_vals, y_vals[fate], '.-', label=fate, color=fate_colors[fate])
        
        
        plt.xlabel(group_param)
        plt.ylabel('Fraction')
        plt.title(f'Fate Fractions vs {group_param}\n'
                  f'm_Planet_eff={eff}, m_Planet/m_J={mp}, a_p/R_Sun={ap}, inc={inc:.2f} rad')
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()
```

```python
plot_group_by_one(df_k1_025, group_param='beta')
```

```python
def plot_compare_eff_vs_ap(df):
    # Set grouping parameter
    group_param = 'a_p/R_Sun'

    # Get other parameters
    _, m_Planet_unique, a_p_unique, beta_unique, inc_unique = get_unique_elements(df)

    # Loop over all combinations except m_Planet_eff and a_p/R_Sun
    for combo in product(m_Planet_unique, beta_unique, inc_unique):
        mp, beta, inc = combo

        fig, ax = plt.subplots(figsize=(8, 5))

        for eff in [0, 1]:
            subdf = df[
                (df['m_Planet_eff'] == eff) &
                (df['m_Planet/m_J'] == mp) &
                (df['beta'] == beta) &
                (df['inc_initial'] == inc)
            ]

            if len(subdf) == 0:
                continue

            # Group by a_p/R_Sun
            grouped = subdf.groupby(group_param)
            x_vals = []
            y_vals = {fate: [] for fate in fate_array}

            for name, group in grouped:
                total = len(group)
                if total == 0:
                    continue
                try:
                    x_vals.append(float(name))  # convert string to float if necessary
                except:
                    x_vals.append(name)
                for fate in fate_array:
                    count = len(group[group['final_fate'] == fate])
                    y_vals[fate].append(count / total)

            # Sort by x
            sorted_idx = np.argsort(x_vals)
            x_vals_sorted = np.array(x_vals)[sorted_idx]
            for fate in fate_array:
                y = np.array(y_vals[fate])[sorted_idx]
                if len(y):
                    label = f'{fate}, eff={eff}'
                    ax.plot(x_vals_sorted, y, '.-' if eff==1 else '.--', label=label, color=fate_colors[fate], alpha=0.8 if eff == 1 else 0.4)

        ax.set_xlabel('a_p/R_Sun')
        ax.set_ylabel('Fraction')
        ax.set_title(f'Fate vs a_p/R_Sun\nm_Planet/m_J={mp}, beta={beta:.3g}, inc={inc:.2f} rad')
        ax.legend()
        ax.grid(True)
        plt.tight_layout()
        plt.show()
```

```python
plot_compare_eff_vs_ap(df_k1_025)
```
